Fokus utama: data cleaning, eksplorasi data, filtering, grouping, dan penanganan missing values.

## 1. Install & Import Library

```python
# Install pandas jika belum ada
# !pip install pandas

import pandas as pd
print('Pandas version:', pd.__version__)
```

## 2. Menampilkan Data (5 Baris Pertama)

```python
# Menampilkan data
df = pd.read_csv('datacovid19.csv')

# Tampilkan 5 baris pertama
print('5 baris pertama dataset:')
print(df.head())
print()
print('Ukuran dataset:', df.shape)
```

## 3. Filtering Data

```python
# Filtering data dari tanggal tertentu
# Pastikan kolom date bertipe string atau datetime
df['date'] = df['date'].astype(str)

# Filter data US pada tanggal 18 Maret 2020
us_data = df[df['country_region'] == 'US']
us_march_data = us_data[us_data['date'] == '18-03-20']

print('Data setelah filtering (US, 18-03-20):')
print(us_march_data)
```

## 4. Grouping Data

```python
# Grouping data berdasarkan country_region dan date
country_date_group = df.groupby(['country_region', 'date']).sum(numeric_only=True)

print('Data setelah grouping berdasarkan country_region dan date:')
print(country_date_group)
print()
print('Ukuran data setelah grouping:', country_date_group.shape)
```

## 5. Cek Data Kosong

```python
# Cek data kosong
print('Jumlah data kosong di setiap kolom:')
print(df.isnull().sum())
```

## 6. Mengisi Nilai Kosong dengan Rata-rata

```python
# Mengisi nilai kosong dengan nilai rata-rata (untuk kolom numerik)
for col in df.select_dtypes(include=['number']).columns:
    df[col] = df[col].fillna(df[col].mean())

print('Data setelah mengisi nilai kosong dengan rata-rata:')
print(df.head(10))
```

## 7. Mengisi Nilai Kosong dengan Nilai yang Sering Muncul (Modus)

```python
# Mengganti data kosong dengan nilai yang sering muncul (untuk kolom kategori/object)
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print('Data setelah mengisi nilai kosong dengan modus:')
print(df.head(10))
```

## 8. Cek Data Kosong Setelah Diperbaiki

```python
# Cek data kosong setelah diperbaiki
print('Jumlah data kosong di setiap kolom setelah diperbaiki:')
print(df.isnull().sum())
```

## 9. Filtering & Grouping Lanjutan

```python
# Membuat data sederhana untuk filtering dan grouping
data = {
    'Country': ['US', 'UK', 'Guam', 'US', 'UK', 'Guam'],
    'Cases':   [50, 150, 200, 10, 300, 20],
    'Deaths':  [1, 10, 0, 2, 15, 1]
}

df2 = pd.DataFrame(data)

# Filtering: Memilih data di mana Cases lebih dari 100
filtered_df = df2[df2['Cases'] > 100]
print('Data setelah filtering (Cases > 100):')
print(filtered_df)

print()

# Grouping: Mengelompokkan data berdasarkan Country dan menghitung total Cases dan Deaths
grouped_df = df2.groupby('Country').sum()
print('\nData setelah grouping berdasarkan Country (total Cases dan Deaths per negara):')
print(grouped_df)
```

## 10. Cek & Isi Data Kosong (Contoh Sederhana)

```python
# Data awal dalam bentuk dictionary dengan nilai kosong
data = {
    'Country': ['US', 'US', 'US', None, 'US', 'US'],
    'Cases':   [126, 85, None, 151, 208, 202],
    'Deaths':  [9, 3, 0, 9, None, 9]
}

df3 = pd.DataFrame(data)

# Cek jumlah data kosong di setiap kolom
print('Jumlah data kosong di setiap kolom:')
print(df3.isnull().sum())

print()

# Isi data kosong pada kolom numerik dengan rata-rata
df3['Cases']  = df3['Cases'].fillna(df3['Cases'].mean())
df3['Deaths'] = df3['Deaths'].fillna(df3['Deaths'].mean())

# Isi data kosong pada kolom objek dengan modus
df3['Country'] = df3['Country'].fillna(df3['Country'].mode()[0])

print('Data setelah diisi:')
print(df3)

print()
print('Jumlah data kosong di setiap kolom setelah diisi:')
print(df3.isnull().sum())
```

- Setelah proses cleaning dengan rata-rata (numerik) dan modus (kategori), semua kolom berhasil dibersihkan dari missing values
